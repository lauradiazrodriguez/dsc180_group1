# Poster Visuals

```python
import matplotlib.pyplot as plt
import numpy as np
```

# Comparing MSEs and Number of Predictors

```python
baseline_mse = 0.89524
specialized_mse = 0.87664

models = ['Baseline PCMCI', 'Specialized PCMCI']
mse_values = [baseline_mse, specialized_mse]

plt.figure(figsize=(8,7))

bars = plt.bar(models, mse_values)

plt.ylabel("Out-of-Sample Predictive Mean Squared Error (MSE)")
plt.title("Predictive Performance: Mean Squared Error (MSE)")

# annotate bars
for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2,
             height,
             f"{height:.4f}",
             ha='center',
             va='bottom')

plt.tight_layout()
plt.show()
```

```python
baseline_preds = 17.5
specialized_preds = 3.8

baseline_mse = 0.89524
specialized_mse = 0.87664

models = ['Baseline PCMCI', 'Specialized PCMCI']
num_vars = [baseline_preds, specialized_preds]
mse_values = [baseline_mse, specialized_mse]

x = np.arange(len(models))
width = 0.35

fig, ax1 = plt.subplots(figsize=(7,5))

# Left axis → number of variables
bars1 = ax1.bar(x - width/2, num_vars, width, label='Number of Predictors Used')
ax1.set_ylabel('Number of Predictors Used')
ax1.set_xticks(x)
ax1.set_xticklabels(models)

# Right axis → MSE
ax2 = ax1.twinx()
bars2 = ax2.bar(x + width/2, mse_values, width, label='Out-of-Sample MSE')
ax2.set_ylabel('MSE')

# Title
plt.title("Baseline vs Domain-Specialized PCMCI")

# Annotate values
for bar in bars1:
    ax1.text(bar.get_x() + bar.get_width()/2,
             bar.get_height(),
             f'{bar.get_height()}',
             ha='center',
             va='bottom')

for bar in bars2:
    ax2.text(bar.get_x() + bar.get_width()/2,
             bar.get_height(),
             f'{bar.get_height():.4f}',
             ha='center',
             va='bottom')

fig.tight_layout()
plt.show()
```

```python
baseline_mse = 0.89524
specialized_mse = 0.87664

baseline_preds = 17.5
specialized_preds = 3.8

models = ['Baseline PCMCI', 'Specialized PCMCI']

mse_values = [baseline_mse, specialized_mse]
predictor_counts = [baseline_preds, specialized_preds]

fig, axes = plt.subplots(1, 2, figsize=(10,4))

# --- MSE comparison ---
axes[0].bar(models, mse_values)
axes[0].set_title("Out-of-Sample Prediction Error")
axes[0].set_ylabel("MSE")

for i, v in enumerate(mse_values):
    axes[0].text(i, v, f"{v:.4f}", ha='center', va='bottom')

# --- Predictor comparison ---
axes[1].bar(models, predictor_counts)
axes[1].set_title("Number of Predictors Required")
axes[1].set_ylabel("Number of Predictors")

for i, v in enumerate(predictor_counts):
    axes[1].text(i, v, f"{v}", ha='center', va='bottom')

plt.tight_layout()
plt.show()
```

```python
baseline_mse = 0.89524
specialized_mse = 0.87664

models = ['Baseline PCMCI', 'Specialized PCMCI']
mse_values = [baseline_mse, specialized_mse]

# Determine colors (lower MSE is better)
colors = ['indianred', 'yellowgreen'] if baseline_mse > specialized_mse else ['yellowgreen', 'indianred']

plt.figure(figsize=(5,4))

bars = plt.bar(models, mse_values, color=colors)

plt.ylabel("Out-of-Sample MSE")
plt.title("Prediction Error Comparison")

# Annotate values
for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2,
             height,
             f"{height:.4f}",
             ha='center',
             va='bottom')

plt.tight_layout()
plt.show()
```

```python
baseline_preds = 17.5
specialized_preds = 3.8

models = ['Baseline PCMCI', 'Specialized PCMCI']
predictor_counts = [baseline_preds, specialized_preds]

# Determine colors (fewer predictors is better)
colors = ['indianred', 'yellowgreen'] if baseline_preds > specialized_preds else ['yellowgreen', 'indianred']

plt.figure(figsize=(5,4))

bars = plt.bar(models, predictor_counts, color=colors)

plt.ylabel("Number of Predictors")
plt.title("Model Complexity Comparison")

# Annotate values
for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2,
             height,
             f"{int(height)}",
             ha='center',
             va='bottom')

plt.tight_layout()
plt.show()
```

```python
baseline_mse = 0.89524
specialized_mse = 0.87664

models = ["Baseline PCMCI", "Fine-Tuned PCMCI"]

# Better/worse semantic colors
better_color = "darkseagreen"
worse_color = "indianred"

# -----------------------------
# Figure 1: Out-of-sample MSE
# Lower is better
# -----------------------------
mse_values = [baseline_mse, specialized_mse]
mse_colors = (
    [worse_color, better_color]
    if baseline_mse > specialized_mse
    else [better_color, worse_color]
)

fig, ax = plt.subplots(figsize=(12, 8))
x = np.arange(len(models))

bars = ax.bar(
    x, mse_values, width=0.62,
    color=mse_colors, linewidth=0.8
)

ax.set_xticks(x)
ax.set_xticklabels(models)
ax.set_ylabel("Out-of-Sample MSE")
ax.set_title("Prediction Error on Held-Out Data", pad=10)

# Light y-grid only
ax.grid(axis="y", linestyle="--", linewidth=0.6, alpha=0.5)
ax.set_axisbelow(True)

# Remove unnecessary spines
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

# Add headroom for labels
ymax = max(mse_values)
ax.set_ylim(0, ymax * 1.18)

# Value labels
for bar, val in zip(bars, mse_values):
    ax.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height() + ymax * 0.025,
        f"{val:.4f}",
        ha="center", va="bottom"
    )

fig.tight_layout()
fig.savefig("mse_comparison_publication.png", bbox_inches="tight")
fig.savefig("mse_comparison_publication.pdf", bbox_inches="tight")
plt.show()
```

```python
baseline_num_preds = 17.5
specialized_num_preds = 3.8

# -----------------------------
# Figure 2: Number of predictors
# Lower is better
# -----------------------------
predictor_values = [baseline_num_preds, specialized_num_preds]
predictor_colors = (
    [worse_color, better_color]
    if baseline_num_preds > specialized_num_preds
    else [better_color, worse_color]
)

fig, ax = plt.subplots(figsize=(12, 8))
x = np.arange(len(models))

bars = ax.bar(
    x, predictor_values, width=0.62,
    color=predictor_colors, linewidth=0.8
)

ax.set_xticks(x)
ax.set_xticklabels(models)

# ---- Make graph box grey ----
for spine in ax.spines.values():
    spine.set_color("#888888")

ax.tick_params(axis='x', pad=10)
ax.tick_params(axis='y', pad=1, colors="#888888")

ax.set_ylabel("Number of Predictors", labelpad=18)
ax.set_title("Model Sparsity", pad=18)

ax.grid(axis="y", linestyle="--", linewidth=0.6, alpha=0.5)
ax.set_axisbelow(True)

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

ymax = max(predictor_values)
ax.set_ylim(0, ymax * 1.18)

for bar, val in zip(bars, predictor_values):
    ax.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height() + ymax * 0.025,
        f"{int(val)}",
        ha="center", va="bottom"
    )

fig.tight_layout()
fig.savefig("predictor_count.pdf", bbox_inches="tight")
ax.grid(False)
plt.show()
```
